**Ablauf:** Ein Neuron reicht für XOR nicht. Wir kombinieren OR, NAND und «Keine Prüfung» zu AND, verändern die Gewichte und sehen sofort Netz, Ausgaben und Entscheidungsgrenzen. Danach trainieren wir ein anderes XOR-Netz.

## Rückblick · Warum eine Gerade nicht reicht
Bei **P = 0** gehören (0,1) und (1,0) zur Ja-Klasse, (0,0) und (1,1) zur Nein-Klasse. Ein einzelnes Neuron hat nur eine gerade Entscheidungsgrenze und kann die vier Punkte nicht trennen.

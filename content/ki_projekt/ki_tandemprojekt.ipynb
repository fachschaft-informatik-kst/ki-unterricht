<details><summary>Wichtige Begriffe</summary>

| Begriff | Bedeutung im Projekt |
|---|---|
| **Label / Klasse** | Die richtige Antwort zu einem Foto, etwa «Apfel» |
| **Exemplar / Gruppe** | Ein einzelner Gegenstand oder eine einzelne Person. Eine Gruppe (T1, T2, V1, P1, P2) enthält genau ein Exemplar je Klasse. |
| **Merkmale** | 1280 Zahlen, mit denen MobileNet jedes Foto beschreibt. Sie bleiben fest. |
| **Neuron, versteckte Schicht** | Unser Netz hat 16 Neuronen, die Merkmale zu Zwischenwerten verrechnen |
| **Epoche** | Ein Durchgang durch alle Trainingsbilder (hier 60) |
| **Startwert (Seed)** | Legt die Zufallsanteile fest, damit ein Lauf wiederholbar ist (hier 11) |
| **Training** | Hier lernt das Netz seine Gewichte (T1, T2). Darf oft angesehen werden. |
| **Validierung** | Hier wird die Variante gewählt (V1). Darf beliebig oft angesehen werden, bis die Wahl feststeht. |
| **Test (Abschlusstest)** | Hier wird die gewählte Variante einmal geprüft (P1, P2). Danach nichts mehr ändern. |
| **Datenleck** | Dasselbe Exemplar steckt in Training und Prüfung. Die Prüfung wird zu leicht. |
| **Trefferquote** | Anteil richtiger Vorhersagen (auch «Genauigkeit»). Bei zwei gleich grossen Klassen sind 50 % reines Raten. |
| **Prozentpunkte** | Differenz zweier Prozentwerte: 75 % − 69 % = 6 Prozentpunkte |
| **Modellwert** | Sicherheit des Netzes für eine Klasse, keine Garantie für Richtigkeit |
| **Beobachtung / Hypothese** | Was am Bild belegbar ist / eine mögliche, noch ungeprüfte Erklärung |
</details>

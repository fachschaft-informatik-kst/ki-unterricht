## Hilfe bei Problemen
- **Nach Reload:** Vorbereitung und Datenimport erneut starten.

- **Test gesperrt:** Erst Auswahl begründen, dann `TEST_FREIGEBEN = True`.
- **Weitere Fehler:** Meldung notieren und mit dem gesicherten Stand gezielt Hilfe holen.

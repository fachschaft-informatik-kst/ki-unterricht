## Phase 4 · Unsere Untersuchung
**Frage:** Welchen Einfluss untersuchen wir? Wählen Sie genau einen Pflichtweg.

| Weg | A | B | Konstant |
|---|---|---|---|
| `menge` | 12 Bilder pro Klasse | 48 Bilder pro Klasse | Trainingspersonenpool, Netz, Epochen, Lernrate, Validierung |
| `vielfalt` | Zwei Trainingspersonen | Alle vier Trainingspersonen | 24 Bilder pro Klasse, Netz, Epochen, Lernrate, Validierung |
| `netz` | 8 versteckte Neuronen | 48 versteckte Neuronen | Exakt gleiche Trainingsbilder, Epochen, Lernrate, Validierung |

Beim Mengenvergleich sind die kleinen Datensätze Teilmengen der grossen. Die Epochenzahl bleibt gleich; mehr Bilder bedeuten dabei auch mehr Verarbeitung pro Epoche. Beim Vielfaltvergleich müssen sich die Bilder bei gleicher Anzahl unterscheiden. Beim Netzvergleich bleiben die ausgewählten Bilder pro Lauf identisch.

### Vor dem Training ausfüllen
**Unsere präzise Frage:** …

**Kontrollpunkt 4:** Versuchsplan am … festgehalten, bevor Ergebnisse sichtbar sind.

```python
WEG = 'vielfalt'  # 'menge', 'vielfalt' oder 'netz'

# Nur die zum gewählten Weg gehörenden Werte bearbeiten, vor dem Versuch festlegen.
BILDER_PRO_KLASSE = 24  # Vielfalt / Netz
MENGEN = (12, 48)       # nur Menge: A und B
WENIGE_PERSONEN = ['P01', 'P02']  # nur Vielfalt: A
NETZGROESSEN = (8, 48)  # nur Netz: A und B
NEURONEN = 24          # Menge / Vielfalt
EPOCHEN = 60
LERNRATE = 0.003
SEEDS = (11, 22, 33)    # alle Läufe berichten
DATEN_SEED = 2026
```

**Experiment:** Nun A und B trainieren. Sechs kleine Trainingsläufe; im Browser kann dies etwas dauern. Einstellungen werden beim Start im Experiment gespeichert. Änderungen oben wirken erst durch erneutes Starten dieser Zelle und erneute Auswertung.

```python
experiment = ki.experiment_starten(
    daten, weg=WEG, bilder_pro_klasse=BILDER_PRO_KLASSE,
    mengen=MENGEN, wenige_personen=WENIGE_PERSONEN,
    netzgroessen=NETZGROESSEN, neuronen=NEURONEN,
    epochen=EPOCHEN, lernrate=LERNRATE, seeds=SEEDS, daten_seed=DATEN_SEED)
ki.ergebnisse_zeigen(experiment)
```

### Belege festhalten

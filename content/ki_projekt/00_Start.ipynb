Dort finden Sie den vollständigen Ablauf, Termine, Themenideen, Bewertungs- und Kompetenzraster sowie die Abgabecheckliste. Sie können Ihren Fortschritt festhalten und sich auf das Fachgespräch vorbereiten.

## Im Notebook arbeiten

1. **[Lernlabor](ki_lernlabor.ipynb):** Training und Auswertung gemeinsam kennenlernen.
2. **[Foto-Projektjournal](ki_tandemprojekt.ipynb):** Eigene Daten importieren, Netz trainieren und Untersuchung dokumentieren. Zuerst mit `UEBUNG` beginnen.
3. **[Symbolprojekt](ki_symbolprojekt.ipynb):** Alternative nach Absprache für selbst gezeichnete Symbole.

Das Fotostudio laden Sie über die vorbereitete Zelle im Fotojournal herunter. Lassen Sie die Projektdateien zusammen in diesem Ordner.

**Speichern:** Notebook mit Ausgaben und Daten regelmässig ausserhalb des Browsers sichern. Die Checkliste auf der Webseite ersetzt Ihre Abgabe nicht.

# KI-Tandemprojekt · Hier starten

In diesem Ordner finden Sie alles für Ihr Projekt. Öffnen Sie die Materialien über die Links in der folgenden Reihenfolge.

| Schritt | Material | Wofür? |
|---|---|---|
| 1 | [Projektauftrag](ki_projektauftrag.md) | Ziel, Termine, Abgabe und Bewertung kennenlernen |
| 2 | [Ideenpool](ki_ideenpool.md) | Thema wählen und Projektsteckbrief vorbereiten |
| 3 | [Lernlabor](ki_lernlabor.ipynb) | Training und Auswertung gemeinsam üben |
| 4 | [Bildsammlung planen](ki_fotodaten.md) | Eigene Fotos und getrennte Datengruppen vorbereiten |
| 5 | **[Projektjournal öffnen](ki_tandemprojekt.ipynb)** | Bilder importieren, Netz trainieren, Versuche dokumentieren |

**Im Projektjournal beginnen Sie mit `UEBUNG`.** Für Ihr eigenes Projekt wechseln Sie danach zu `EIGENE_DATEN`. Das Fotostudio laden Sie direkt über die vorbereitete Zelle im Journal herunter.

**Speichern:** Sichern Sie Ihr ausgefülltes Notebook mit Ausgaben und Ihre Daten regelmässig ausserhalb des Browsers. Lassen Sie die Projektdateien in diesem Ordner zusammen, damit die vorbereiteten Funktionen und Übungsdaten gefunden werden.

**Alternative nach Absprache:** [Symbolprojekt](ki_symbolprojekt.ipynb) für selbst gezeichnete Symbole.

Die Python-, JavaScript-, HTML- und JSON-Dateien sind vorbereitete Hilfsmittel. Sie müssen sie für den Einstieg nicht einzeln öffnen.

# Unser Foto-Projekt
**Zu zweit · zwei Bildklassen · ein fairer Vergleich**

> Helfen vielfältigere Trainingsbilder beim Erkennen neuer Gegenstände?

1. [Lernlabor](ki_lernlabor.ipynb): den Ablauf mit Übungsdaten ausprobieren.
2. [Projektjournal](ki_tandemprojekt.ipynb): eigene Fotos, Vergleich und fünf Reflexionen.
3. [Projektseite](https://fachschaft-informatik-kst.github.io/ki-unterricht/projekt/): Themen, Datenplan, Termine und Bewertung.

**Abgabe:** Journal mit Ausgaben, Fotos/JSON und Versuchsprotokoll. Fachgespräch gemeinsam im Tandem.

**Datenplan:** 80 Fotos = 2 Klassen × 5 unabhängige Gruppen × 8 Fotos.
Die Fotos möglichst vor dem 15.12. sammeln. Ersatzdaten nach Absprache.

Die Dateien in diesem Ordner zusammen belassen. `ki_symbolprojekt.ipynb` ist eine anspruchsvollere Zusatzroute nach Absprache, kein zweiter Pflichtauftrag.

Ein künstliches Neuron erhält zwei Eingaben (0 oder 1). Es multipliziert sie mit Gewichten, addiert den Bias und gibt 1 aus, wenn das Ergebnis mindestens 0 ist. **Vermutet zuerst**, wann die Ausgabe 1 ist.
 

### Zuerst überlegen, dann den Code ausführen

Stellt euch zwei Fragen vor, auf die jemand nur mit **Ja (1)** oder **Nein (0)** antworten kann. Eine Entscheidung hängt zum Beispiel davon ab, ob die Sonne scheint und ob eine Freundin Zeit hat. Der **Eingang** des Neurons sind zwei Zahlen, **Gewichte** verändern den Einfluss der Eingaben, und der **Bias** verschiebt die Grenze für ein Ja.

Ein Beispiel: Bei den Gewichten 1 und 1 und dem Bias −1,5 erhält (1,0) die Summe 1 + 0 − 1,5 = −0,5. Das Ergebnis ist 0, weil die Summe kleiner als 0 ist. Bei (1,1) ergibt sich 0,5 und somit 1. Das ist die UND-Regel (AND). Ihr könnt die vier Fälle mit eigenen Zahlen nachrechnen.

**Wichtig:** In diesem Notebook werden die Gewichte und der Bias **von euch verändert**, nicht vom Computer gelernt. Erst wenn ein Lernverfahren passende Werte aus Beispielen berechnet, sprechen wir von *Training*. Diese Unterscheidung braucht ihr später im Bildprojekt.

**Arbeitsweise:** Vor jeder Änderung zuerst notieren, welche Zeilen sich vermutlich verändern. Dann die Zelle erneut mit **Shift + Enter** ausführen, die tatsächlichen Ausgaben vergleichen und auf der Kursseite in eigenen Worten erklären. Wenn ein Versuch nicht wie erwartet funktioniert, dokumentiert den Unterschied. Ein richtiger Wert ohne Begründung zeigt noch nicht, dass ihr das Modell verstanden habt.


```python
def neuron(x1, x2, w1=1, w2=1, bias=-1.5):
    summe = x1*w1 + x2*w2 + bias
    return int(summe >= 0)

faelle = [(0,0),(0,1),(1,0),(1,1)]
for x1,x2 in faelle:
    print((x1,x2), '→', neuron(x1,x2))
```

## Versuch 1: AND oder OR?
Ändert **nur** `bias` von `-1.5` auf `-0.5`. Führt die nächste Zelle erneut aus. Was ändert sich?

```python
bias = -1.5  # Ändern auf -0.5
for x1,x2 in faelle:
    print((x1,x2), '→', neuron(x1,x2,bias=bias))
```

## Versuch 2: Ein Gewicht ändern
Stellt den Bias wieder auf `-1.5` und ändert **nur** `w1`. Welche Eingaben sind nun entscheidend?

```python
w1 = 1
for x1,x2 in faelle:
    print((x1,x2), '→', neuron(x1,x2,w1=w1,bias=-1.5))
```

**Sicherung:** Notiert in der Kursseite eine Einstellung für AND, eine für OR und erklärt den Einfluss des Bias in einem Satz.

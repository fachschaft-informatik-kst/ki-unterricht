Ein künstliches Neuron erhält zwei Eingaben (0 oder 1). Es multipliziert sie mit Gewichten, addiert den Bias und gibt 1 aus, wenn das Ergebnis mindestens 0 ist. **Vermutet zuerst**, wann die Ausgabe 1 ist.

```python
def neuron(x1, x2, w1=1, w2=1, bias=-1.5):
    summe = x1*w1 + x2*w2 + bias
    return int(summe >= 0)

faelle = [(0,0),(0,1),(1,0),(1,1)]
for x1,x2 in faelle:
    print((x1,x2), '→', neuron(x1,x2))
```

## Versuch 1: AND oder OR?
Ändert **nur** `bias` von `-1.5` auf `-0.5`. Führt die nächste Zelle erneut aus. Was ändert sich?

```python
bias = -1.5  # Ändern auf -0.5
for x1,x2 in faelle:
    print((x1,x2), '→', neuron(x1,x2,bias=bias))
```

## Versuch 2: Ein Gewicht ändern
Stellt den Bias wieder auf `-1.5` und ändert **nur** `w1`. Welche Eingaben sind nun entscheidend?

```python
w1 = 1
for x1,x2 in faelle:
    print((x1,x2), '→', neuron(x1,x2,w1=w1,bias=-1.5))
```

**Sicherung:** Notiert in der Kursseite eine Einstellung für AND, eine für OR und erklärt den Einfluss des Bias in einem Satz.

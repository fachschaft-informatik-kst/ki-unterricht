## Aufbauen · Drei Rollen für Daten  ⏱ 4 Min.
Eine Prüfung zeigt, warum es **drei** Rollen braucht:

| Rolle | Vergleich | Frage | Gewichte ändern? | Modellwahl beeinflussen? | Ansehen |
|---|---|---|---|---|---|
| **Training** | Übungsaufgaben mit Lösung | Was lernt das Netz? | ja | nein | beliebig oft |
| **Validierung** | Probeprüfung | Welche Variante ist besser? | nein | **ja** | beliebig oft, solange wir wählen |
| **Test** | echte Prüfung | Wie gut ist die gewählte Variante bei wirklich neuen Exemplaren? | nein | nein | **genau einmal**, am Ende |

**Warum nicht nur zwei Rollen?** Wer eine Probeprüfung immer wieder ansieht und danach nachbessert, passt sich ihr an. Ihr Ergebnis ist dann zu rosig. Der Test muss deshalb aus Daten bestehen, die bei **keiner** Entscheidung beteiligt waren. Wer nach dem Test nachbessert, macht ihn zur zweiten Validierung.

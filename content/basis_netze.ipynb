XOR bedeutet: **genau eine** der beiden Eingaben ist 1. Ein einzelnes Schwellwert-Neuron kann XOR nicht für alle vier Fälle abbilden. Zwei Zwischenschritte helfen.


### Warum mehrere Neuronen?

Die XOR-Regel bedeutet **entweder das eine oder das andere, aber nicht beide**. Sie liefert für (0,0) eine 0, für (0,1) und (1,0) jeweils eine 1 und für (1,1) wieder eine 0. Vergleicht diese vier Zielwerte mit eurer vorherigen AND/OR-Tabelle.

Bei einem einzigen Schwellwert-Neuron versucht ihr, durch zwei Gewichte und den Bias genau diese vier Ergebnisse einzustellen. Ihr könnt manche Zeilen treffen, aber nicht alle gleichzeitig: Ein einzelnes solches Neuron zieht eine **gerade Entscheidungsgrenze**. Die vier XOR-Fälle können damit nicht vollständig getrennt werden. Diesen mathematischen Beweis müsst ihr noch nicht führen; prüft die Behauptung anhand eurer Versuche.

Die **versteckte Schicht** kann zwei Zwischenfragen stellen. Das erste Neuron prüft OR: «Ist mindestens eine Eingabe 1?» Das zweite prüft NAND: «Sind *nicht beide* Eingaben 1?» Die Ausgabeschicht kombiniert die Antworten wiederum mit AND. Nur wenn OR und NAND beide 1 liefern, erhalten wir XOR. Zeichnet den Weg für die Eingabe (1,1) nach und vergleicht ihn mit (1,0).

**Achtung:** Dieses Netz hat von uns fest vorgegebene Gewichte. Es zeigt, **was mehrere Neuronen gemeinsam ausdrücken können**, aber es trainiert noch nicht selbstständig. Im Bildprojekt werden dagegen Gewichte anhand von Trainingsdaten angepasst. 

**Euer Lernprodukt:** Ihr könnt die Rolle von Eingabeschicht, Zwischenschicht und Ausgabeschicht an einem konkreten Fall erklären. Ihr könnt ausserdem begründen, weshalb mehr Schichten nicht automatisch bessere Entscheidungen bedeuten: Es kommt darauf an, welche Regeln beziehungsweise Merkmale die Neuronen umsetzen.


```python
faelle = [(0,0),(0,1),(1,0),(1,1)]
def neuron(a,b,w1,w2,bias):
    return int(a*w1+b*w2+bias >= 0)
for a,b in faelle:
    print((a,b),'XOR-Ziel:',int(a!=b))
```

## Versuch 1: Ein einzelnes Neuron
Verändert Gewichte und Bias. Schafft ihr alle vier XOR-Fälle?

```python
w1,w2,bias = 1,1,-0.5
for a,b in faelle:
    print((a,b),'Ziel',int(a!=b),'Neuron',neuron(a,b,w1,w2,bias))
```

## Versuch 2: Ein Netz mit einer Zwischenschicht
Die versteckten Neuronen berechnen OR und NAND. Das Ausgabeneuron verknüpft sie durch AND. **Vermutet** die Ausgabe für (1,1).

```python
def xor_netz(a,b):
    oder = neuron(a,b,1,1,-0.5)
    nand = neuron(a,b,-1,-1,1.5)
    ausgabe = neuron(oder,nand,1,1,-1.5)
    return oder,nand,ausgabe
for a,b in faelle:
    print((a,b),'OR, NAND, XOR:',xor_netz(a,b))
```
